```python
import re

def reformat_markdown_tasks(input_filepath, output_filepath):
    with open(input_filepath, 'r', encoding='utf-8') as f:
        lines = f.read().split('\n')

    out_lines = []

    # Regex patterns to identify tasks, subtasks, and bullet points
    task_regex = re.compile(r'^### Task (\d+)')
    subtask_regex = re.compile(r'^#### Subtask (\d+)\.\d+')
    bullet_regex = re.compile(r'^\s*\*\s+')

    in_target_section = False
    current_task = 0
    current_subtask = 0
    bullet_count = 0

    for line in lines:
        # Check for a new Task
        task_match = task_regex.match(line)
        if task_match:
            current_task = int(task_match.group(1))
            # Only apply the fixing logic for Task 15 and above
            in_target_section = current_task >= 15
            out_lines.append(line)
            continue
            
        if in_target_section:
            # Check for an existing Subtask header
            subtask_match = subtask_regex.match(line)
            if subtask_match:
                # Increment our internal subtask counter and reset the bullet count
                current_subtask += 1
                bullet_count = 0
                out_lines.append(f"#### Subtask {current_task}.{current_subtask}")
                continue
                
            # Check for a bullet point
            if bullet_regex.match(line):
                bullet_count += 1
                if bullet_count > 1:
                    # We found a second bullet point, so we split into a new subtask
                    current_subtask += 1
                    
                    # Ensure there is a blank line before the new header for clean formatting
                    if out_lines and out_lines[-1].strip() != "":
                        out_lines.append("")
                        
                    out_lines.append(f"#### Subtask {current_task}.{current_subtask}")
                    
                    # This bullet point is now the first bullet of the newly created subtask
                    bullet_count = 1
                    
            out_lines.append(line)
        else:
            # We are before Task 15, so just append the line exactly as it was
            out_lines.append(line)

    # Write the formatted output to a new file
    with open(output_filepath, 'w', encoding='utf-8') as f:
        f.write('\n'.join(out_lines))
        
    print(f"Success! Processed markdown saved to: {output_filepath}")
```

```python

# --- Execution ---
# Replace these with your actual file names
input_file = '../reference/retrieval_tasks.md' 
output_file = '../reference/retrieval_tasks_test.md'

reformat_markdown_tasks(input_file, output_file)
```
